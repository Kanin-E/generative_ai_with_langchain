# Built-in middleware

Prebuilt middleware for common agent use cases


LangChain provides prebuilt middleware for common use cases. Each middleware is production-ready and configurable for your specific needs.

1 Middleware can be at many hook places


# Provider-agnostic middleware

The following middleware work with any LLM provider:

Middleware	Description
- <b>Summarization</b>:	Automatically summarize conversation history when approaching token limits.
- <b>Human-in-the-loop</b>:	Pause execution for human approval of tool calls.
- <b>Model call limit</b>:	Limit the number of model calls to prevent excessive costs.
- <b>Tool call limit</b>:	Control tool execution by limiting call counts.
- <b>Model fallback</b>:	Automatically fallback to alternative models when primary fails.
- <b>PII detection</b>:	Detect and handle Personally Identifiable Information (PII).
- <b>To-do list</b>:	Equip agents with task planning and tracking capabilities.
- <b>LLM tool selector</b>:	Use an LLM to select relevant tools before calling main model.
- <b>Tool retry</b>:	Automatically retry failed tool calls with exponential backoff.
- <b>Model retry</b>:	Automatically retry failed model calls with exponential backoff.
- <b>LLM tool emulator</b>:	Emulate tool execution using an LLM for testing purposes.
- <b>Context editing</b>:	Manage conversation context by trimming or clearing tool uses.
- <b>Shell tool</b>:	Expose a persistent shell session to agents for command execution.
- <b>File search</b>:	Provide Glob and Grep search tools over filesystem files.
