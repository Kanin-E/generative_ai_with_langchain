# Messages types

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

# Option1
conversation = [
    {"role": "system", "content": "You are a helpful assistant that translates English to French."},
    {"role": "user", "content": "Translate: I love programming."},
    {"role": "assistant", "content": "J'adore la programmation."},
    {"role": "user", "content": "Translate: I love building applications."}
]

response = model.invoke(conversation)
print(response.content)
```

# Methods of model

```python
conversation = [
    SystemMessage("You are a comedian with sarcastic style, like to nag people in a funny way"),
    HumanMessage("Tell me a brief biography of Harry potter")
]
```

consists of 3 main methods (or more but i don't know)
- <b>invoke()</b>: processes a single input synchronously and returns an output. simple method return AI message
- <b>stream()</b>: Streams output as it's being generated, for better user experience seeing the output generated real time. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. <b>return an iterator </b>
- <b>batch()</b>: Efficiently processes multiple inputs in parallel, Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:
- <b>ainvoke(), abatch(). astream()</b>: Asymchronous version of the above methods

## Invoke

simple method return AI message. The most straightforward way to call a model is to use invoke() with a single message or a list of messages.

```python
response = model.invoke(conversation)
print('Return type :',type(response))
print('')
print('=====================================')
print(response.content)
```

## Stream

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.
Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
responses = model.stream(conversation)
print('Return type :',type(response))
print('')
print('=====================================')
for chunk in responses:
    print(chunk)
```

```python
responses = model.stream(conversation)
print('Return type :',type(response))
print('')
print('=====================================')
for chunk in responses:
    print(chunk.content)
```

```python
responses = model.stream(conversation)
print('Return type :',type(response))
print('')
print('=====================================')
for chunk in responses:
    print(chunk.text)
```

## Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response)
```
