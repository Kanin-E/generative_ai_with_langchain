Runnable is any component that can process inputs and produce outputs in a standardized way. Every component built with LCEL adheres to this interface, which provides consistent methods including:

- <b>invoke()</b>: processes a single input synchronously and returns an output
- <b>stream()</b>: Streams output as it's being generated
- <b>batch()</b>: Efficiently processes multiple inputs in parallel
- <b>ainvoke(), abatch(). astream()</b>: Asymchronous version of the above methods


any runnable components can be connected to any other runnable


make it simpler to use to build a sophisicated workflow Use pipe operator |

the langchain take advantage of python operator overloading concept and redefine the __or__ function to connect each Runnable togethers

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

chat = ChatGoogleGenerativeAI(model="gemini-2.5-pro")

# First chain generates a story
story_prompt = PromptTemplate.from_template("Write a short story about {topic}")
story_chain = story_prompt | chat | StrOutputParser()

# Second chain analyzes the story
analysis_prompt = PromptTemplate.from_template("Analyze the following story's mood:\n{story}")
analysis_chain = analysis_prompt | chat | StrOutputParser()

output_prompt = PromptTemplate.from_template(
    "Here's the story: \n{story}\n\nHere's the mood: \n{mood}"
)
# Combine chains
story_with_analysis = story_chain | analysis_chain

# Run the combined chain
result = story_with_analysis.stream({"topic": "a rainy day"})
for line in result:
    print(line)
```
